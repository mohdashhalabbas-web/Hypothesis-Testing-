##Hypothesis Testing
Question 1: What is a null hypothesis (H₀) and why is it important in hypothesis testing?

Answer 1:  A null hypothesis (H₀) is a statement that there is no significant difference, effect, or relationship in a population. It is the default assumption in hypothesis testing.

The null hypothesis is important because it provides a baseline for comparison. Statistical tests help determine whether there is enough evidence to reject H₀ or fail to reject H₀.

Example: H₀: There is no significant difference between the average scores of two groups.


Question 2: What does the significance level (α) represent in hypothesis testing?

Answe 2: The significance level (α) is the probability of rejecting the null hypothesis (H₀) when it is actually true. It represents the maximum acceptable risk of making a Type I error.

Common significance levels are 0.05 (5%), 0.01 (1%), and 0.10 (10%).

For example, if α = 0.05, there is a 5% risk of incorrectly rejecting a true null hypothesis.

Question 3: Differentiate between Type I and Type II errors.

Answer 3: Type I Error: A Type I error occurs when we reject the null hypothesis (H₀) even though it is actually true. The probability of making a Type I error is equal to the significance level (α).

Type II Error: A Type II error occurs when we fail to reject the null hypothesis (H₀) even though it is actually false. Its probability is represented by β.

In simple terms:
Type I Error = Rejecting a true H₀.
Type II Error = Failing to reject a false H₀.

Question 4: Explain the difference between a one-tailed and two-tailed test. Give an example of each.

Answer 4: A one-tailed test is used when the alternative hypothesis predicts a specific direction, such as greater than or less than. The rejection region is located in only one tail of the distribution.

Example:
H₀: The average score is equal to 50.
H₁: The average score is greater than 50.
This is a one-tailed test.

A two-tailed test is used when the alternative hypothesis checks for a difference in either direction, without specifying whether it is higher or lower. The rejection regions are located in both tails of the distribution.

Example:
H₀: The average score is equal to 50.
H₁: The average score is not equal to 50.
This is a two-tailed test.

In simple terms:
One-tailed test = Tests for an effect in one specific direction.
Two-tailed test = Tests for an effect in either direction.

Question 5: A company claims that the average time to resolve a customer complaint is 10 minutes.
 A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes.

At α = 0.05, test the claim.

Answer 5:

Given:
Population mean (μ₀) = 10 minutes
Sample mean (x̄) = 12 minutes
Sample standard deviation (s) = 3 minutes
Sample size (n) = 9
Significance level (α) = 0.05

Step 1: State the hypotheses

H₀: μ = 10 minutes
H₁: μ ≠ 10 minutes

Since the claim is about equality, this is a two-tailed t-test.

Step 2: Calculate the t-statistic

t = (x̄ - μ₀) / (s / √n)

t = (12 - 10) / (3 / √9)
t = 2 / 1
t = 2.00

Degrees of freedom = n - 1 = 8

Step 3: Critical value

At α = 0.05 and 8 degrees of freedom, the critical t-value is approximately ±2.306.

Step 4: Decision

Calculated t-value = 2.00
Critical t-value = ±2.306

Since 2.00 < 2.306, we fail to reject the null hypothesis (H₀).

Conclusion:

There is not enough statistical evidence at the 5% significance level to conclude that the average complaint resolution time is different from 10 minutes.

Therefore, the company's claim that the average resolution time is 10 minutes is not rejected.

Question 6: When should you use a Z-test instead of a t-test?

Answer 6:  A Z-test is generally used when the population standard deviation (σ) is known or when the sample size is large (usually n ≥ 30). It is based on the standard normal (Z) distribution.

A t-test is used when the population standard deviation is unknown and the sample size is small (usually n < 30). It uses the Student's t-distribution.

In simple terms:

Z-test → Population standard deviation is known or sample size is large.
t-test → Population standard deviation is unknown and sample size is small.

Example:
If a sample of 50 observations is taken and the population standard deviation is known, a Z-test can be used.
If a sample of 15 observations is taken and the population standard deviation is unknown, a t-test should be used.



Question 7: The productivity of 6 employees was measured before and after a training program
(Table)
At α = 0.05, test if the training improved productivity.

Answer 7:

Given:

Before = [50, 60, 58, 55, 62, 56]
After = [55, 65, 59, 58, 63, 59]
α = 0.05

Step 1: Calculate the differences

Difference = After - Before

Differences = [5, 5, 1, 3, 1, 3]

Mean difference = 3
Standard deviation of differences = 2
Sample size (n) = 6

Step 2: State the hypotheses

H₀: There is no improvement in productivity (μd = 0)
H₁: Training improves productivity (μd > 0)

This is a one-tailed paired t-test.

Step 3: Calculate the t-statistic

t = Mean difference / (Standard deviation / √n)

t = 3 / (2 / √6)
t = 3.674

Degrees of freedom = n - 1 = 5

Step 4: Critical value

At α = 0.05 and 5 degrees of freedom, the critical t-value is approximately 2.015.

Step 5: Decision

Calculated t-value = 3.674
Critical t-value = 2.015

Since 3.674 > 2.015, we reject H₀.

Conclusion:

There is sufficient statistical evidence at the 5% significance level to conclude that the training program significantly improved employee productivity.



Question 8: A company wants to test if product preference is independent of gender (Table) At α = 0.05, test independence

Answer 8:

Given:

Male: Product A = 30, Product B = 20
Female: Product A = 10, Product B = 40
Total = 100
α = 0.05

Step 1: State the hypotheses

H₀: Product preference is independent of gender.
H₁: Product preference is not independent of gender.

Step 2: Calculate the expected frequencies

Expected frequency = (Row Total × Column Total) / Grand Total

Male, Product A = (50 × 40) / 100 = 20
Male, Product B = (50 × 60) / 100 = 30
Female, Product A = (50 × 40) / 100 = 20
Female, Product B = (50 × 60) / 100 = 30

Step 3: Calculate the Chi-square statistic

χ² = Σ (Observed - Expected)² / Expected

χ² = (30-20)²/20 + (20-30)²/30 + (10-20)²/20 + (40-30)²/30

χ² = 16.67

Degrees of freedom = (2-1)(2-1) = 1

Step 4: Critical value

At α = 0.05 and 1 degree of freedom, the critical value is 3.841.

Step 5: Decision

Calculated χ² = 16.67
Critical χ² = 3.841

Since 16.67 > 3.841, we reject H₀.

Conclusion:

There is sufficient statistical evidence at the 5% significance level to conclude that product preference is dependent on gender.

Therefore, product preference and gender are significantly associated.
